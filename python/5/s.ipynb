```python
from abc import ABC, abstractmethod
class animal(ABC):
    @abstractmethod
    def move(self):
        pass
class bird(animal):
    def move(self):
        print("bird")
class cat(animal):
    def move(self):
        print("cat")
a = bird()
a.move()  
```

```python
class A():
    def do_this(self):
        print("A")
class B(A):
    pass
class C():
    def do_this(self):
        print("C")
class D(B,C):
    pass
d = D()
d.do_this()  
```
